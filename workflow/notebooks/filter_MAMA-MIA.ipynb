# Download MAMA-MIA from Synapse

```python
from synapseclient import Synapse, File
from synapseclient.operations import get
from synapseclient.models import Project, Folder
from synapseutils.sync import syncFromSynapse, generateManifest, generate_sync_manifest
from synapseutils.walk_functions import walk
```

```python
syn = Synapse()
syn.login()

image_dir_id = "syn64871114"
duke1_id = "syn64883023"

walk_results = walk(syn, image_dir_id, includeTypes=['file'])

# list(walk_results)

for dirpath, dirnames, filenames in walk_results:
    
    for sample in filenames:
        file_id = str(sample[0])
        if file_id.endswith('0000.nii.gz'):
            # print(sample)
            patient_id = file_id.removesuffix('_0000.nii.gz')
            out_path = dirs.RAWDATA / dataset / "images/scans" / patient_id

            syn_id = sample[1]
            syncFromSynapse(syn, entity=syn_id, path=out_path, ifcollision='keep.local', manifest='root')

# generateManifest(image_dir_id, allFiles filename=dirs.RAWDATA / dataset / "metadata/images_manifest.tsv")
```
